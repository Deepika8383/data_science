## 🧩 Data Encoding

Data encoding is the process of converting categorical data into numerical format so that machine learning algorithms can understand and process it.

---

### 🔹 1. Label Encoding

- Assigns each unique category a numeric value.
- Best for **ordinal data** (e.g., Low, Medium, High).

✅ Pros: Simple and memory-efficient.

⚠️ Cons: Implies an order that may not exist

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Sample DataFrame
data = {
    'Fruit': ['Apple', 'Banana', 'Mango', 'Apple', 'Banana', 'Mango']
}
df = pd.DataFrame(data)

# Initialize LabelEncoder
le = LabelEncoder()

# Fit and transform the 'Fruit' column
df['Fruit_encoded'] = le.fit_transform(df['Fruit'])

# Show result
print(df)
```

### 🔹 2. One-Hot Encoding

Creates separate binary columns for each category.

Best for nominal data (no natural order).

✅ Pros: No ordinal assumptions

⚠️ Cons: Increases dimensionality

```python
import pandas as pd

# Sample DataFrame
data = {
    'Fruit': ['Apple', 'Banana', 'Mango', 'Apple', 'Banana', 'Mango']
}
df = pd.DataFrame(data)

# Apply one-hot encoding
df_encoded = pd.get_dummies(df, columns=['Fruit'])

# Show result
print(df_encoded)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Sample DataFrame
df = pd.DataFrame({'Fruit': ['Apple', 'Banana', 'Mango', 'Apple']})

# Initialize OneHotEncoder with correct parameter
encoder = OneHotEncoder(sparse_output=False)  # Use 'sparse_output' instead of 'sparse'

# Fit and transform
encoded = encoder.fit_transform(df[['Fruit']])

# Create a DataFrame with encoded values
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['Fruit']))

# Concatenate original and encoded DataFrame
result = pd.concat([df, encoded_df], axis=1)

print(result)
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Sample data
df = pd.DataFrame({'Fruit': ['Apple', 'Banana', 'Mango', 'Apple']})

# Use sparse_output=False to get dense array
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[['Fruit']])

# Convert to DataFrame
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['Fruit']))

# Combine original + encoded
result = pd.concat([df, encoded_df], axis=1)
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
import sys
from scipy.sparse import issparse

# Create a sample dataframe with many repeated categories
df = pd.DataFrame({'Category': ['A'] * 1_000 + ['B'] * 1_000 + ['C'] * 1_000})

# OneHotEncoder with sparse matrix (default)
encoder_sparse = OneHotEncoder(sparse_output=True)
sparse_encoded = encoder_sparse.fit_transform(df[['Category']])

# OneHotEncoder with dense array
encoder_dense = OneHotEncoder(sparse_output=False)
dense_encoded = encoder_dense.fit_transform(df[['Category']])

# Print info about the encoding
print("✅ Sparse matrix shape:", sparse_encoded.shape)
print("✅ Dense array shape:", dense_encoded.shape)

# Memory usage comparison
sparse_size = sparse_encoded.data.nbytes + sparse_encoded.indptr.nbytes + sparse_encoded.indices.nbytes
dense_size = dense_encoded.nbytes

print(f"\n📦 Memory used by Sparse Matrix: {sparse_size / 1024:.2f} KB")
print(f"📦 Memory used by Dense Array: {dense_size / 1024:.2f} KB")

# Optional: convert to DataFrame for visual inspection
dense_df = pd.DataFrame(dense_encoded, columns=encoder_dense.get_feature_names_out(['Category']))
print("\n🔍 Preview of Dense One-Hot Encoded DataFrame:")
print(dense_df.head())
```

## Ordinal Encoding
Ordinal Encoding assigns integer values to categories based on a meaningful order or rank.

- Useful when categories have an inherent order (e.g., Low, Medium, High).
- Encodes categories as 0, 1, 2, ... preserving the order.
- Not suitable for nominal categories without order.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Sample data with order
df = pd.DataFrame({'Size': ['Small', 'Medium', 'Large', 'Small', 'Large']})

# Define the order explicitly
size_order = [['Small', 'Medium', 'Large']]

# Initialize and fit the encoder
ordinal_encoder = OrdinalEncoder(categories=size_order)
ordinal_encoded = ordinal_encoder.fit_transform(df[['Size']])

# Add encoded values to DataFrame
df['Size_Encoded'] = ordinal_encoded

print("🔢 Ordinal Encoding:")
print(df)
```

## Binary Encoding
Binary Encoding converts categorical variables into binary digits. It is useful for high cardinality features by reducing the number of columns compared to one-hot encoding.

- Each category is first assigned an integer.
- Then the integer is converted into binary code.
- Each binary digit becomes a new feature.

```python
import pandas as pd
import category_encoders as ce

# Sample data with high cardinality
df = pd.DataFrame({'City': ['Delhi', 'Mumbai', 'Chennai', 'Kolkata', 'Delhi']})

# Apply Binary Encoding
binary_encoder = ce.BinaryEncoder(cols=['City'])
binary_encoded = binary_encoder.fit_transform(df)

print("🧮 Binary Encoding:")
print(binary_encoded)
```

## Target Guided Ordinal Encoding 
It is a technique used to encode categorical variables based on their relationship with the target variable. This encoding technique is useful when we have a categorical variable with a large number of unique categories, and we want to use this variable as a feature in our machine learning model.

In Target Guided Ordinal Encoding, we replace each category in the categorical variable with a numerical value based on the mean or median of the target variable for that category. This creates a monotonic relationship between the categorical variable and the target variable, which can improve the predictive power of our model.

```python
import pandas as pd

# create a sample dataframe with a categorical variable and a target variable
df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 300, 250, 180, 320]
})

## calculate the mean price for each city
mean_price=df.groupby('city')['price'].mean().to_dict()
print(mean_price)
## replace each city with its mean price
df['city_encoded']=df['city'].map(mean_price)
print(df)
```
