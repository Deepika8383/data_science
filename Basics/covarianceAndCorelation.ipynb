## Covariance and Correlation in Machine Learning

### Covariance
- **Definition:** Covariance measures how two variables change together.
- If both variables tend to increase or decrease together, covariance is **positive**.
- If one variable tends to increase while the other decreases, covariance is **negative**.
- If the variables are independent, covariance is close to **zero**.
- **Formula:**  
  Cov(X,Y)\=1n−1∑i\=1n(Xi−Xˉ)(Yi−Yˉ)\\text{Cov}(X, Y) = \\frac{1}{n - 1} \\sum\_{i=1}^{n} (X\_i - \\bar{X})(Y\_i - \\bar{Y})
- **Interpretation:**
  - Large positive covariance means strong positive linear relationship.
  - Large negative covariance means strong negative linear relationship.
  - Covariance magnitude depends on the scale of variables, so it's hard to interpret directly.

### Correlation
- **Definition:** Correlation is a normalized version of covariance that measures the strength and direction of a linear relationship between two variables.
- It ranges from **-1** to **+1**:
  - +1 means perfect positive linear correlation,
  - -1 means perfect negative linear correlation,
  - 0 means no linear correlation.
- **Formula:**  
  \[
  \text{Corr}(X, Y) = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y}
  \]
  where \(\sigma_X\) and \(\sigma_Y\) are standard deviations of \(X\) and \(Y\).
- **Interpretation:**
  - Correlation is **scale-invariant** and easier to interpret than covariance.
  - Helps identify which features are strongly related in machine learning.

### Why are they important in Machine Learning?
- Both help understand relationships between features.
- Identify redundant or highly correlated features which might affect model performance.
- Important for feature selection, dimensionality reduction (e.g., PCA), and understanding data patterns.


# 📊 Pearson vs Spearman Correlation

Correlation measures the strength and direction of a relationship between two variables.

---

## 📌 Pearson Correlation

- **Type:** Measures **linear** relationships.
- **Scale:** -1 to +1
- **Assumes:** Normal distribution, continuous variables.
- **Sensitive to:** Outliers.

### Formula
r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2} \sqrt{\sum (y_i - \bar{y})^2}}

---

## 📈 Spearman Rank Correlation

The **Spearman Rank Correlation Coefficient** measures the **monotonic relationship** between two variables using their **ranked values** rather than raw data.

---

## 🔍 When to Use

- When data is **not normally distributed**
- When the relationship is **monotonic** but not necessarily linear
- When data is **ordinal** or contains **outliers**

---

## 🧠 Intuition

- Instead of comparing actual values, Spearman compares the **rankings** of those values.
- If two variables increase together monotonically (not necessarily linearly), the correlation is **+1**.
- If one increases while the other decreases monotonically, it's **-1**.

---

## 🧮 Formula

Let \( d_i \) be the **difference between the ranks** of each observation:

```math
\rho = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}


````python
import seaborn as sns
import pandas as pd

# Load dataset
df = sns.load_dataset('healthexp')

# Display first few rows
print(df.head())

# Select only numeric columns
numeric_df = df.select_dtypes(include='number')
print("--------------------------------")
# Compute covariance
print(numeric_df.cov())
print("--------------------------------")
print("Pearson")
print(numeric_df.corr(method='pearson'))
print("--------------------------------")
print("spearman")
print(numeric_df.corr(method='spearman'))
````
