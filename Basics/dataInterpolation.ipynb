## 📈 Data Interpolation

**Data Interpolation** is the technique of estimating unknown values within the range of a dataset using existing data points.

It's useful for:
- Filling in missing data points
- Smoothing or estimating values between measurements


### 🔹 1. Linear Interpolation with NumPy

Linear interpolation assumes the values change linearly between points.

```python
import numpy as np
import matplotlib.pyplot as plt

# Original data
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# New x values for interpolation
x_new = np.linspace(1, 5, 10)
y_interp = np.interp(x_new, x, y)

# Plot original data
plt.scatter(x, y, color='blue', label='Original Data', marker='o')

# Plot interpolated data
plt.scatter(x_new, y_interp, color='red', label='Interpolated Data', marker='x')

# Add labels and legend
plt.title('Linear Interpolation Example')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```


### 📉 2. Cubic Interpolation (with SciPy)

Cubic interpolation fits a **smooth curve** through the data using cubic polynomials. It’s smoother than linear interpolation and works well when data has curves.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# Original data (y = x^3)
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 8, 27, 64, 125])

# Create a cubic interpolation function
f = interp1d(x, y, kind='cubic')

# Interpolated x values
x_new = np.linspace(1, 5, 10)
y_interp = f(x_new)

# Plot original data
plt.scatter(x, y, color='blue', label='Original Data', marker='o')

# Plot interpolated data
plt.scatter(x_new, y_interp, color='red', label='Cubic Interpolation', marker='x')

# Optional: Connect interpolated points with a line for clarity
plt.plot(x_new, y_interp, color='red', linestyle='--', alpha=0.5)

# Add labels and legend
plt.title('Cubic Interpolation Example')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

### 📈 3. Polynomial Interpolation (with NumPy)
Polynomial interpolation fits a **single polynomial** equation (of a chosen degree) through the data points.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create sample data (y = x^2)
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 4, 9, 16, 25])

# Polynomial interpolation (2nd degree)
p = np.polyfit(x, y, 2)  # Fit a polynomial of degree 2
x_new = np.linspace(1, 5, 10)  # New x values
y_interp = np.polyval(p, x_new)  # Evaluate the polynomial at new x values

# Plot original data
plt.scatter(x, y, color='blue', label='Original Data', marker='o')

# Plot interpolated data
plt.scatter(x_new, y_interp, color='green', label='Polynomial Interpolation', marker='x')

# Optional: Connect interpolated points with a line
plt.plot(x_new, y_interp, color='green', linestyle='--', alpha=0.5)

# Add labels and legend
plt.title('Polynomial Interpolation (Degree 2)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
