# 🧠 AI vs ML vs DL vs DS – Easy to Understand Notes


## 📌 1. Artificial Intelligence (AI)

> **AI** is the **broadest** concept. It's the science of making machines **simulate human intelligence** – like learning, reasoning, problem-solving, perception, and language understanding.

- **Goal**: Build **smart systems** that can perform tasks normally requiring human intelligence.
- **Examples**: Chatbots, recommendation systems, self-driving cars.



## 📌 2. Machine Learning (ML)

> **ML** is a **subset of AI** that allows machines to **learn from data** and improve from experience without being explicitly programmed.

- **Goal**: Create models that **learn patterns from data** and make predictions/decisions.
- **Types**:
  - **Supervised Learning** (with labeled data)
  - **Unsupervised Learning** (no labels)
  - **Reinforcement Learning** (rewards-based)

## 🧪 Train, Validation, and Test Sets

When building ML models, we divide data into 3 parts:

### ✅ 1. Training Set
- Used to **train the model**.
- The model learns patterns from this data.

### 📊 2. Validation Set
- Used during training to **tune hyperparameters**.
- Helps avoid **overfitting**.
- Optional: Sometimes combined with cross-validation.

### 🧾 3. Test Set
- Used **after training** to evaluate model performance.
- Ensures that the model generalizes well to **unseen data**.

```python
# Example: Splitting data using sklearn
from sklearn.model_selection import train_test_split

## 🎢 Overfitting vs Underfitting

| Concept        | Description                          | Symptoms                         |
|----------------|--------------------------------------|----------------------------------|
| **Overfitting** | Model learns **too well**, including noise | High train accuracy, low test accuracy |
| **Underfitting** | Model is **too simple** to learn patterns | Low accuracy on both train and test |

### 🛠 How to Fix
- **Overfitting**:
  - Reduce model complexity
  - Add regularization
  - Get more training data
- **Underfitting**:
  - Increase model complexity
  - Train longer
  - Use better features

```python
# Example: Regularization to avoid overfitting
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train, y_train)


## ⚖️ Bias vs Variance Trade-off

- **Bias**: Error due to **simplified assumptions** → underfitting  
- **Variance**: Error due to **model sensitivity to data fluctuations** → overfitting

| Error Type       | Description                                  |
|------------------|----------------------------------------------|
| **High Bias**     | Model is too simple → misses patterns         |
| **High Variance** | Model is too complex → memorizes noise       |




> 🎯 **Goal**: Find the sweet spot between bias and variance for best generalization.

## 📊 Percentiles and Quartiles

### Percentiles
- Percentiles are values that divide a dataset into 100 equal parts.
- The **k-th percentile** is the value below which **k%** of the data falls.
- For example, the 25th percentile is the value below which 25% of the data lies.

### Quartiles
- Quartiles divide data into 4 equal parts (each 25% of data).
- They are specific percentiles:
  - **Minimum**: The smallest value in the data (0th percentile)
  - **First Quartile (Q1)**: 25th percentile — 25% of data falls below this value
  - **Median (Q2)**: 50th percentile — the middle value that divides data into two halves
  - **Third Quartile (Q3)**: 75th percentile — 75% of data falls below this value
  - **Maximum**: The largest value in the data (100th percentile)


### Why important?
- Quartiles and percentiles help understand the **distribution** and **spread** of data.
- They are useful for detecting **outliers**, summarizing data, and comparing datasets.

---

### Visual summary:


````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Sample data
data = [2, 4, 7, 10, 12, 15, 18, 20, 22, 25]

# Calculate quartiles
q1 = np.percentile(data, 25)
median = np.median(data)
q3 = np.percentile(data, 75)
_min = np.min(data)
_max = np.max(data)

plt.figure(figsize=(10, 4))

# Boxplot (horizontal)
sns.boxplot(x=data, color='lightblue')

# Overlay stripplot (data points)
sns.stripplot(x=data, color='darkblue', size=8, jitter=True)

# Annotate quartiles and min/max
plt.text(_min, 0.7, f'Min\n{_min}', horizontalalignment='center', color='red', fontsize=10)
plt.text(q1, 0.7, f'Q1\n{q1}', horizontalalignment='center', color='green', fontsize=10)
plt.text(median, 0.7, f'Median\n{median}', horizontalalignment='center', color='purple', fontsize=10)
plt.text(q3, 0.7, f'Q3\n{q3}', horizontalalignment='center', color='green', fontsize=10)
plt.text(_max, 0.7, f'Max\n{_max}', horizontalalignment='center', color='red', fontsize=10)

# Draw vertical lines for quartiles and min/max
for point in [_min, q1, median, q3, _max]:
    plt.axvline(point, color='gray', linestyle='--', alpha=0.5)

plt.title("Quartiles and Percentiles Visualization")
plt.xlabel("Data values")
plt.yticks([])  # Hide y-axis ticks for cleanliness
plt.show()
````
