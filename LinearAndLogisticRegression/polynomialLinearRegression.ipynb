# 📘 Polynomial Regression

## 📌 Introduction

**Polynomial Regression** is a form of regression analysis in which the relationship between the independent variable \( x \) and the dependent variable \( y \) is modeled as an \( n \)-th degree polynomial.

It extends **Linear Regression** by allowing for non-linear relationships between variables while still using a linear model in terms of the coefficients.

---

## 🧮 Mathematical Equation

A polynomial regression model of degree \( n \) is represented as:

$$
y = \beta_0 + \beta_1x + \beta_2x^2 + \beta_3x^3 + \dots + \beta_nx^n + \varepsilon
$$

Where:
- $( y )$ = Dependent variable  
- $( x )$ = Independent variable  
- $( \beta_0, \beta_1, ..., \beta_n )$ = Coefficients  
- $( \varepsilon )$ = Error term  

> Note: Though this equation is non-linear in terms of \( x \), it's linear in the coefficients, which allows use of linear regression techniques.

---

## 🎯 When to Use

- When data shows a **non-linear** relationship.
- When simple linear regression underfits the data.
- When curve fitting is required for better predictive performance.

---

## 🧠 Assumptions

Polynomial Regression shares the assumptions of linear regression:

1. **Independence**: Observations are independent.
2. **Homoscedasticity**: Constant variance of residuals.
3. **Normality**: Residuals are normally distributed.
4. **Low multicollinearity** (especially important when adding polynomial terms).

---

## 📊 Visualization

Polynomial regression fits a **curved line** to the data. The curve becomes more flexible as the degree of the polynomial increases.

- Degree 1: Linear regression (straight line)
- Degree 2: Quadratic (parabola)
- Degree 3: Cubic curve
- Higher degrees: More complex curves (risk of overfitting)
