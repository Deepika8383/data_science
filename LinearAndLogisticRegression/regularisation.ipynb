# 📉 Regularized Regression Techniques

Regularized regression helps prevent **overfitting** by adding a penalty to the loss function. These techniques are especially useful when:

- There are many predictors (features).
- Predictors are highly correlated (multicollinearity).
- You want to perform feature selection.

---

## 1️⃣ L1 Regression — Lasso (Least Absolute Shrinkage and Selection Operator)

### 📌 Definition
Lasso adds a penalty equal to the **absolute value** of the magnitude of coefficients to the loss function.

### 🧮 Cost Function
$$
\text{Loss}_{L1} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{p}|\beta_j|
$$

Where:
- $( \lambda )$ is the regularization strength.
- $( \beta_j )$ are model coefficients.

### ✅ Advantages
- Performs **feature selection** by shrinking some coefficients to exactly **zero**.
- Good for sparse models.

### ⚠️ Disadvantages
- Can be unstable when predictors are highly correlated.

---

## 2️⃣ L2 Regression — Ridge

### 📌 Definition
Ridge regression adds a penalty equal to the **square** of the magnitude of coefficients.

### 🧮 Cost Function
$$
\text{Loss}_{L2} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{p} \beta_j^2
$$

### ✅ Advantages
- Helps with **multicollinearity** by shrinking coefficients.
- Keeps all predictors but reduces their impact.

### ⚠️ Disadvantages
- Does not eliminate (set to zero) any coefficients → no feature selection.

---

## 3️⃣ Elastic Net Regression

### 📌 Definition
Elastic Net is a **hybrid** of Lasso and Ridge regression. It combines both L1 and L2 penalties.

### 🧮 Cost Function
$$
\text{Loss}_{EN} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 + \lambda_1 \sum_{j=1}^{p}|\beta_j| + \lambda_2 \sum_{j=1}^{p} \beta_j^2
$$

Or often written using a mixing parameter \( \alpha \in [0, 1] \):
$$
\text{Loss}_{EN} = \sum (y_i - \hat{y}_i)^2 + \lambda \left[ \alpha \sum |\beta_j| + (1 - \alpha) \sum \beta_j^2 \right]
$$

### ✅ Advantages
- Works well when there are many correlated features.
- Provides **sparse solutions** like Lasso and **stability** like Ridge.

### ⚠️ Disadvantages
- Needs tuning of two parameters: $( \lambda )$ and $( \alpha )$.

---

## 🧠 Summary Table

| Method       | Penalty Type        | Feature Selection | Best For                     |
|--------------|---------------------|-------------------|------------------------------|
| Lasso (L1)   | $( \sum \beta )$   | ✅ Yes             | Sparse models, feature elimination |
| Ridge (L2)   | $( \sum \beta^2 )$   | ❌ No              | Handling multicollinearity   |
| Elastic Net  | Mix of L1 and L2     | ✅ Partial         | High-dimensional, correlated features |

---

## 🔧 When to Use

- **Lasso**: When you suspect **only a few features** are useful.
- **Ridge**: When **all features** contribute but you want to **shrink** them to avoid overfitting.
- **Elastic Net**: When you want a **balance** of both Lasso and Ridge.
