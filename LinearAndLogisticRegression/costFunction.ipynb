# 🧮 Cost Functions in Regression

Cost functions are used to measure how well a regression model performs by quantifying the difference between the predicted values \( \hat{y} \) and the actual values \( y \). The goal is to minimize these errors to improve model accuracy.

---

## 1️⃣ Mean Squared Error (MSE)

### 📌 Definition
Mean Squared Error is the average of the **squared differences** between actual and predicted values.

### 🧮 Formula
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

### ✅ Advantages
- Penalizes large errors more than small ones due to squaring.
- Smooth gradient, useful for optimization algorithms like Gradient Descent.

### ⚠️ Disadvantages
- Sensitive to outliers because errors are squared.

---

## 2️⃣ Mean Absolute Error (MAE)

### 📌 Definition
Mean Absolute Error is the average of the **absolute differences** between actual and predicted values.

### 🧮 Formula
$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n}|y_i - \hat{y}_i|
$$

### ✅ Advantages
- More robust to outliers compared to MSE.
- Easier to interpret (in same units as the target variable).

### ⚠️ Disadvantages
- Not differentiable at zero, which can be tricky for gradient-based optimization.

---

## 3️⃣ Root Mean Squared Error (RMSE)

### 📌 Definition
Root Mean Squared Error is the square root of the MSE. It brings the error back to the original scale of the target variable.

### 🧮 Formula
$$
\text{RMSE} = \sqrt{ \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 }
$$

### ✅ Advantages
- Same unit as the target variable.
- Penalizes large errors heavily like MSE.

### ⚠️ Disadvantages
- Still sensitive to outliers due to squaring.

---

## 📊 Summary Comparison

| Metric | Formula | Sensitive to Outliers? | Differentiable? | Scale |
|--------|---------|------------------------|------------------|-------|
| MSE    | $( \frac{1}{n} sum (y - \hat{y})^2 )$ | ✅ Yes           | ✅ Yes           | Squared units |
| MAE    | $( \frac{1}{n} sum (y - \hat{y} )$   | 🚫 No            | ⚠️ No at 0       | Same as target |
| RMSE   | $( \sqrt{\text{MSE}} )$                | ✅ Yes           | ✅ Yes           | Same as target |

---

## ✅ Best Practice

- Use **MSE** and **RMSE** when large errors are particularly undesirable.
- Use **MAE** when robustness to outliers is important.
- Compare multiple metrics for a full performance picture.
