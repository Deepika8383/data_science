# 📊 Performance Metrics in Regression

In regression analysis, evaluating the model’s performance is crucial to ensure that it accurately represents the relationship between the dependent and independent variables. Two widely used metrics are **R-squared (R²)** and **Adjusted R-squared**.

---

## 1️⃣ R-squared (R²)

### 📌 Definition

**R-squared**, also known as the **coefficient of determination**, measures the **proportion of the variance in the dependent variable that is predictable from the independent variables**.

### 🧮 Formula

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$

Where:
- $( SS_{res} )$ = Sum of squares of residuals = $( \sum (y_i - \hat{y}_i)^2 )$ 
- $( SS_{tot} )$ = Total sum of squares = $( \sum (y_i - \bar{y})^2 )$

### 📈 Interpretation

- $( R^2 = 0 )$: The model explains **none** of the variability.
- $( R^2 = 1 )$: The model explains **all** the variability.
- $( 0 < R^2 < 1 )$: Indicates the proportion of variance explained.

> 🔹 Example: An R² of 0.80 means **80% of the variance** in the target variable is explained by the model.

### ⚠️ Limitation

- R² **always increases** when you add more predictors—even if they are irrelevant.
- Does **not penalize** for model complexity.

---

## 2️⃣ Adjusted R-squared

### 📌 Definition

**Adjusted R-squared** adjusts the R² value **based on the number of predictors** in the model. It penalizes the inclusion of irrelevant variables and only increases if the new term improves the model more than expected by chance.

### 🧮 Formula

$$
\text{Adjusted } R^2 = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)
$$

Where:
- \( n \) = Number of observations  
- \( p \) = Number of predictors (independent variables)  

### 📈 Interpretation

- Can **decrease** if new predictors do not improve the model.
- Helps in **model selection** by balancing goodness-of-fit and complexity.

> 🔹 Preferred over R² when comparing models with a different number of predictors.

---

## 📊 Summary Comparison

| Metric           | Formula                                                      | Penalizes Complexity | Value Range | Use Case                                |
|------------------|--------------------------------------------------------------|-----------------------|-------------|------------------------------------------|
| R-squared (R²)   | $( 1 - \frac{SS_{res}}{SS_{tot}} )$                          | ❌ No                 | 0 to 1      | Measure of explained variance            |
| Adjusted R²      | $( 1 - \frac{(1 - R^2)(n - 1)}{n - p - 1} )$                 | ✅ Yes                | ≤ R²        | Comparing models with different predictors |

---

## ✅ Best Practices

- Use **R²** to understand overall variance explained.
- Use **Adjusted R²** when building or comparing models with multiple predictors.
- Combine these with **residual analysis** and **cross-validation** for robust evaluation.
