# 📘 Logistic Regression

## 🔍 What is Logistic Regression?

**Logistic Regression** is a supervised machine learning algorithm used for **classification** tasks. It predicts the **probability** of an input belonging to a certain class (typically binary: 0 or 1).

---

## ⚙️ How Does It Work?

1. Takes input features (e.g., age, income).
2. Applies a **linear equation**:  
   
   $z = w_0 + w_1x_1 + w_2x_2 + \dots + w_nx_n$
   
3. Passes `z` through a **sigmoid (logistic) function**:
   
   $h(x) = \frac{1}{1 + e^{-z}}$
4. Uses a **threshold** (usually 0.5):
   - If output ≥ 0.5 → Class 1
   - If output < 0.5 → Class 0

---

## ✅ Use Cases

- Email Spam Detection
- Medical Diagnosis (e.g., tumor prediction)
- Customer Churn Prediction
- Credit Card Fraud Detection

---

## 🔑 Advantages

- Simple and easy to implement
- Works well for binary classification
- Outputs probabilities (confidence)

---

## ⚠️ Limitations

- Assumes linear relationship between features and log-odds
- Not ideal for complex nonlinear relationships without transformations

---

# ❓ Why Not Use Linear Regression for Classification?

### 🚫 Problems with Linear Regression:

1. **Unbounded Outputs**  
   - Linear regression can output values like -1.3 or 2.5 ❌  
   - Classification requires probabilities between 0 and 1 ✅

2. **No Probabilistic Interpretation**  
   - Logistic regression gives meaningful probabilities (e.g., 85% chance of spam)
   - Linear regression does not

3. **Sensitive to Outliers**  
   - Can lead to incorrect boundaries for class labels

4. **Inaccurate Decision Boundary**  
   - Logistic regression provides a **sigmoid curve**, giving a **smooth** and **bounded** transition between classes

---

### 🟢 Example

| Age | Bought Product? (0 = No, 1 = Yes) |
|-----|----------------|
| 22  | 0              |
| 25  | 0              |
| 30  | 1              |
| 35  | 1              |

- **Linear Regression Prediction:**
  - Age 24 → 0.3  
  - Age 38 → 1.6 ❌ (invalid probability)

- **Logistic Regression Prediction:**
  - Age 24 → 0.25 ✅  
  - Age 38 → 0.95 ✅

---

## 🧠 Conclusion

> **Linear regression is not suitable for classification** because:
> - It does **not restrict outputs to 0–1**
> - It lacks **probabilistic meaning**
> - It produces **poor decision boundaries**
>  
> **Logistic regression** solves these issues using the **sigmoid function**.

---


# 📉 Logistic Regression Cost Function

## ❓ Why Do We Need a Cost Function?

In logistic regression, we predict **probabilities**, and we need a cost function that:
- Penalizes wrong predictions,
- Works well with the **sigmoid** output (non-linear),
- And helps the model learn by minimizing the error.

---

## 🚫 Why Not Use Mean Squared Error (MSE)?

MSE is used in linear regression, but it's not suitable here because:
- It results in a **non-convex cost function** for logistic regression.
- That makes it **hard for gradient descent** to converge.

---

## ✅ Logistic Loss / Binary Cross-Entropy

The cost function used in logistic regression is:

\[
J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(h_\theta(x^{(i)})) + (1 - y^{(i)}) \log(1 - h_\theta(x^{(i)})) \right]
\]

Where:
- \( m \): number of training examples  
- \( h_\theta(x) \): predicted probability (sigmoid output)  
- \( y \): actual label (0 or 1)

---

## 🧠 Intuition

- If $( y = 1 )$, the cost is:  
  
  $-\log(h_\theta(x))$

  → penalizes small probabilities near 0

- If $( y = 0 )$, the cost is:  
  
  $-\log(1 - h_\theta(x))$
  
  → penalizes large probabilities near 1

```python
from sklearn.datasets import load_iris
```

```python
dataset=load_iris()
print(dataset.DESCR)
```

```python
dataset.keys()
```

```python
import pandas as pd
import numpy as np
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
df['target']=dataset.target
df.columns
```

```python
## Binary Classification
df_copy=df[df['target']!=2]
```

```python
X=df_copy.iloc[:,:-1]
y=df_copy.iloc[:,-1]
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=20, random_state=42)
```

```python
classifier=LogisticRegression()
```

```python
classifier.fit(X_train,y_train)
```

```python
classifier.predict_proba(X_test)
```

```python
## Prediction
y_pred=classifier.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```

## Hyperparameter Tuning

```python
## Gridsearchcv
from sklearn.model_selection import GridSearchCV
import warnings
warnings.filterwarnings('ignore')
```

```python
parameters={'penalty':('l1','l2','elasticnet',None),'C':[1,10,20]}
```

```python
clf=GridSearchCV(classifier,param_grid=parameters,cv=5)
```

## Logistic Regression For Multiclass Classification

```python
# make a prediction with a multinomial logistic regression model
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
# define dataset
X, y = make_classification(n_samples=1000, n_features=10, n_informative=5, n_redundant=5, n_classes=3, random_state=1)
```

```python
# define the multinomial logistic regression model For multiclass classification
model = LogisticRegression(multi_class='ovr', solver='lbfgs')
```

```python
# Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
# trianing model
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
model.predict_proba(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```
