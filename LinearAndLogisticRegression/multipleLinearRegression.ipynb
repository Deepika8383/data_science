# 📘 Multiple Linear Regression (MLR)

## 📌 Introduction

**Multiple Linear Regression (MLR)** is a statistical technique used to model the relationship between one **dependent variable** and **two or more independent variables**. It helps in understanding how changes in the independent variables are associated with changes in the dependent variable.

MLR is a fundamental concept in statistics and machine learning, widely used in forecasting, predictive modeling, and inferential analysis.

---

## 🧮 Mathematical Model

$$
Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + \dots + \beta_nX_n + \varepsilon
$$

Where:
- $( Y )$ = Dependent variable (response)
- $( X_1, X_2, ..., X_n )$ = Independent variables (predictors)
- $( \beta_0 )$ = Intercept (value of Y when all Xs are 0)
- $( \beta_1, ..., \beta_n )$ = Coefficients (effect size of each X on Y)
- $( \varepsilon )$ = Error term (captures variability not explained by the model)

---

## 🎯 Objective

The goal of MLR is to estimate the coefficients \( \beta_0, \beta_1, ..., \beta_n \) such that the predicted values of \( Y \) are as close as possible to the observed values. This is typically done using the **least squares method**, which minimizes the sum of squared residuals.

---

## 🧠 Key Assumptions

To ensure reliable and interpretable results, MLR makes several assumptions:

1. **Linearity**: The relationship between the dependent and each independent variable is linear.
2. **Independence of errors**: Residuals (errors) are independent of each other.
3. **Homoscedasticity**: Constant variance of residuals across all levels of independent variables.
4. **Normality of errors**: Residuals are normally distributed.
5. **No multicollinearity**: Independent variables are not highly correlated with each other.

---

## 🔍 Interpretation of Coefficients

- **Intercept (β₀)**: Expected value of Y when all Xs = 0.
- **Slope coefficients (β₁ to βₙ)**: Represent the change in Y for a one-unit increase in the corresponding X, keeping other Xs constant.

---

## 📊 Model Evaluation Metrics

- **R² (Coefficient of Determination)**: Proportion of variance in Y explained by the model.
- **Adjusted R²**: Adjusted for the number of predictors; better for multiple regression.
- **Mean Squared Error (MSE)**: Average of squared residuals.
- **Root Mean Squared Error (RMSE)**: Square root of MSE, interpretable in original units.
- **Mean Absolute Error (MAE)**: Average of absolute residuals.

---

## ⚠️ Multicollinearity

Occurs when independent variables are highly correlated with each other. It can inflate the variance of coefficient estimates and make the model unstable.

### Detection:
- **Variance Inflation Factor (VIF)**: VIF > 5 or 10 indicates multicollinearity.

---

## 🧪 Example Use Cases

- Predicting house prices using features like size, location, number of rooms, etc.
- Forecasting sales based on advertising budget, seasonality, and product prices.
- Estimating student performance using study hours, attendance, and test scores.

## 🛠️ Implementation Tools

- **Python**: `scikit-learn`, `statsmodels`
- **R**: `lm()` function

---

## ✅ Best Practices

- Always visualize data before modeling.
- Check assumptions using residual plots and diagnostics.
- Use feature selection techniques to reduce dimensionality.
- Regularize the model (e.g., Ridge, Lasso) if multicollinearity is present.
