# 📈 Simple Linear Regression

## 🔍 What is Simple Linear Regression?

Simple Linear Regression is a **supervised machine learning** algorithm that models the relationship between:
- **One independent variable (X)** — also called the predictor or feature
- **One dependent variable (Y)** — also called the target or outcome

It assumes a **linear relationship** between X and Y.

---

## 📘 Equation of a Line

The mathematical equation for simple linear regression is:


$Y = \beta_0 + \beta_1 X + \varepsilon$



Where:
- $( Y )$: Target variable (dependent)
- $( X )$: Feature variable (independent)
- $(\beta_0)$: Intercept (value of Y when X = 0)
- $( \beta_1 )$: Slope of the line (change in Y per unit change in X)
- $( \varepsilon)$: Error term (residual)

---

## 🛠️ How It Works

1. The algorithm finds the **best-fitting straight line** through the data points.
2. It minimizes the **sum of squared residuals** (difference between actual and predicted Y values).

---

## 🧮 Cost Function (Error)

To measure how well the model is performing, we use a **Cost Function** — specifically, the **Mean Squared Error (MSE)**:


$J(m, c) = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

Where:
- $n$ is the number of training examples
- $y_i$ is the actual value
- $\hat{y}_i$ is the predicted value = $(mx_i + c)$

> The goal is to minimize this cost function.

---

## 🔁 Gradient Descent Algorithm

**Gradient Descent** is an optimization algorithm used to minimize the cost function by updating the model parameters ($m$, $c$) iteratively.

### 🔧 Update Rules


$m := m - \alpha \cdot \frac{\partial J}{\partial m}$

$c := c - \alpha \cdot \frac{\partial J}{\partial c}$

Where:
- $α$ (alpha) is the **learning rate**
- The partial derivatives (gradients) are:


$\frac{\partial J}{\partial m} = -\frac{1}{n} \sum_{i=1}^{n} x_i (y_i - \hat{y}_i)$


$\frac{\partial J}{\partial c} = -\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)$


---

## 📉 Convergence Algorithm

The Gradient Descent algorithm stops when the model **converges**, i.e., the change in cost becomes very small or zero.

### 🔁 Convergence Criteria:

- Stop when:

$|J_{old} - J_{new}| < \epsilon$

Where:
- $ε$ is a small number like `1e-6`
- $J_old$ and $J_new$ are the cost values from two consecutive iterations

---

## 🧠 Summary

- We start with initial guesses for `m` and `c`.
- We compute the cost using MSE.
- We update `m` and `c` using gradient descent.
- We repeat until the cost converges (i.e., doesn’t change much anymore).
- Once converged, the model has learned the best-fit line.

---

## 📌 Key Terms

| Term             | Description                                       |
|------------------|---------------------------------------------------|
| Cost Function     | Measures prediction error                        |
| Gradient Descent  | Optimizes parameters to reduce error             |
| Learning Rate     | Controls size of each step                       |
| Convergence       | When further updates cause negligible change     |
